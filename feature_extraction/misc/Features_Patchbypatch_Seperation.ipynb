```python
import os
import h5py
import torch
import numpy as np
from tqdm import tqdm

# Paths
input_root = r"D:\Aamir Gulzar\KSA_project2\dataset\Features\Conch_v1_5"
output_root = r"D:\Aamir Gulzar\KSA_project2\dataset\Features\Conch_v1_5_patchbypatch"

os.makedirs(output_root, exist_ok=True)

# Process each slide folder
for slide_name in tqdm(os.listdir(input_root), desc="🔍 Processing slides"):
    slide_path = os.path.join(input_root, slide_name, "patch_features")
    h5_path = os.path.join(slide_path, f"{slide_name}_patch_features.h5")
    
    if not os.path.isfile(h5_path):
        print(f"⚠️ H5 file not found for {slide_name}")
        continue

    try:
        with h5py.File(h5_path, 'r') as f:
            coords = f["coords"][:].squeeze(0)     # Shape: (N, 2)
            features = f["features"][:].squeeze(0)  # Shape: (N, D)

            if coords.shape[0] != features.shape[0]:
                print(f"❌ Mismatch in number of coords and features for {slide_name}")
                continue

            # Output folder for this slide
            slide_output_dir = os.path.join(output_root, slide_name)
            os.makedirs(slide_output_dir, exist_ok=True)

            # Save each patch's feature as a .pt file
            for i in range(features.shape[0]):
                x, y = coords[i]
                feature_tensor = torch.tensor(features[i], dtype=torch.float32)

                pt_name = f"{slide_name}_x{int(x)}_y{int(y)}.pt"
                pt_path = os.path.join(slide_output_dir, pt_name)

                torch.save(feature_tensor, pt_path)

    except Exception as e:
        print(f"❌ Error processing {slide_name}: {e}")
```
