### This script will create a csv that will match each CRC100K patch to its respective label.

```python
import os
import csv

# ====== SET YOUR PATHS ======
root_folder = r"D:\Aamir Gulzar\dataset\CRC100K\NCT-CRC-HE-100K-NONORM"
output_csv = r"D:\Aamir Gulzar\KSA_project2\Cancer-detection-classifier\slide_aggregation\caption_generation\CRC100K_ground_truth_labels.csv"
# ============================

rows = []

# Loop over label folders
for label in os.listdir(root_folder):
    label_path = os.path.join(root_folder, label)

    if os.path.isdir(label_path):  # ensure it's a folder
        for patch_name in os.listdir(label_path):
            patch_path = os.path.join(label_path, patch_name)

            if os.path.isfile(patch_path):  # ensure it's a file
                rows.append([patch_name, label])

# Write CSV
with open(output_csv, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["patch_name", "label"])  # header
    writer.writerows(rows)

print("CSV created at:", output_csv)
```
